# Baseline Model Results

This notebook displays the evaluation results for the logistic regression baseline model trained on the SECOM semiconductor manufacturing dataset. The model uses `class_weight="balanced"` to handle the ~6.6% fail-rate class imbalance and is evaluated on a held-out 20% stratified test split. All artifacts (metrics JSON, figures, feature importance CSV) are pre-computed by the pipeline scripts and loaded here for display only.

```python
import json
from pathlib import Path

metrics = json.loads(Path('../reports/baseline_metrics.json').read_text())
print(f"ROC AUC:   {metrics['roc_auc']:.3f}")
print(f"PR AUC:    {metrics['pr_auc']:.3f}")
print(f"Precision: {metrics['precision']:.3f}")
print(f"Recall:    {metrics['recall']:.3f}")
print(f"F1 Score:  {metrics['f1']:.3f}")
```
